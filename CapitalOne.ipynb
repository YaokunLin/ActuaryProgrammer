```python
# https://www.1point3acres.com/bbs/thread-1178703-1-1.html
# Q4

def solution(numRows, numColumns, curRow, curColumn, laserCoordinates):

    # Treat positions outside the board as virtual boundaries
    max_row_up = 0
    min_row_down = numRows + 1
    max_col_left = 0
    min_col_right = numColumns + 1

    horizontal_blocked = False
    vertical_blocked = False

    for laser_row, laser_col in laserCoordinates:

        # Rows affect vertical movement
        if laser_row < curRow:
            max_row_up = max(max_row_up, laser_row)
        elif laser_row > curRow:
            min_row_down = min(min_row_down, laser_row)
        else:
            # Entire current row is covered
            horizontal_blocked = True

        # Columns affect horizontal movement
        if laser_col < curColumn:
            max_col_left = max(max_col_left, laser_col)
        elif laser_col > curColumn:
            min_col_right = min(min_col_right, laser_col)
        else:
            # Entire current column is covered
            vertical_blocked = True

    up = (
        0 if vertical_blocked
        else curRow - max_row_up - 1
    )

    down = (
        0 if vertical_blocked
        else min_row_down - curRow - 1
    )

    left = (
        0 if horizontal_blocked
        else curColumn - max_col_left - 1
    )

    right = (
        0 if horizontal_blocked
        else min_col_right - curColumn - 1
    )

    return max(up, down, left, right)

numRows = 8
numColumns = 8
curRow = 5
curColumn = 3
laserCoordinates = [[1, 6], [2, 8]]
solution(numRows, numColumns, curRow, curColumn, laserCoordinates)
```

```python
def solution(numRows, numColumns, curRow, curColumn, laserCoordinates):
    laser_rows = set()
    laser_cols = set()

    for r, c in laserCoordinates:
        laser_rows.add(r)
        laser_cols.add(c)

    directions = [
        (-1, 0),  # up
        (1, 0),   # down
        (0, -1),  # left
        (0, 1)    # right
    ]

    res = 0

    for dr, dc in directions:
        r, c = curRow, curColumn
        count = 0

        while True:
            r += dr
            c += dc

            # outside board
            if r < 1 or r > numRows or c < 1 or c > numColumns:
                break

            # destroyed by a laser
            if r in laser_rows or c in laser_cols:
                break

            count += 1

        res = max(res, count)

    return res

numRows = 8
numColumns = 8
curRow = 5
curColumn = 3
laserCoordinates = [[1, 6], [2, 8]]
solution(numRows, numColumns, curRow, curColumn, laserCoordinates)
```

```python
# https://www.1point3acres.com/interview/thread/1190124
# Q2

def solution(grid, pattern):
  n_rows, n_cols = len(grid), len(grid[0])
  n_p_rows, n_p_cols = len(pattern), len(pattern[0])

  def is_match(r, c):
    for p_r in range(n_p_rows):
      for p_c in range(n_p_cols):
        if r + p_r >= n_rows or c + p_c >= n_cols:
          return False
        if pattern[p_r][p_c] != grid[r + p_r][c + p_c]:
          return False
    return True

  for r in range(n_rows):
    for c in range(n_cols):
      if grid[r][c] == pattern[0][0] and is_match(r, c):
        return True
      else:
        continue
  return False


grid = [
    [1, 2, 4, 5],
    [2, 3, 6, 7],
    [6, 4, 2, 2]
]

pattern = [
    [6, 7],
    [2, 3]
]

solution(grid, pattern)
```
