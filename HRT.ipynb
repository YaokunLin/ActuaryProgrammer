```python
import pandas as pd
import numpy as np

np.random.seed(42)

# -------------------------
# 1. Settings
# -------------------------
tickers = ["AAPL", "MSFT", "NVDA", "AMZN", "GOOG", "JPM"]

dates = pd.bdate_range(
    start="2026-01-02",
    periods=100
)

starting_prices = np.array([
    250,   # AAPL
    420,   # MSFT
    180,   # NVDA
    230,   # AMZN
    195,   # GOOG
    270    # JPM
])

# Daily expected returns
mu = np.array([
    0.0005,
    0.0004,
    0.0008,
    0.0005,
    0.0005,
    0.0003
])

# Daily volatility
vol = np.array([
    0.018,
    0.016,
    0.030,
    0.021,
    0.019,
    0.015
])

corr = np.array([
    [1.00, 0.70, 0.65, 0.60, 0.65, 0.35],
    [0.70, 1.00, 0.65, 0.60, 0.70, 0.40],
    [0.65, 0.65, 1.00, 0.55, 0.60, 0.25],
    [0.60, 0.60, 0.55, 1.00, 0.60, 0.30],
    [0.65, 0.70, 0.60, 0.60, 1.00, 0.35],
    [0.35, 0.40, 0.25, 0.30, 0.35, 1.00]
])

cov = np.outer(vol, vol) * corr

returns = np.random.multivariate_normal(
    mean=mu,
    cov=cov,
    size=len(dates)
)

returns_df = pd.DataFrame(
    returns,
    index=dates,
    columns=tickers
)

prices_df = (1 + returns_df).cumprod() * starting_prices

prices_df = prices_df.round(2)

prices_df.head(10)
```
